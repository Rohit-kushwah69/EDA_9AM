**lambda function**

- lambda function is a function we can represnt in a single line

- it behave exatcly how our origianl function

- lambda arguments : expression

**case-1: function with one argument**

```python
def add(n):
    return(n+10)
add(20)
```

```python
# syntax
<function name>= lambda <arg> : <return value>
# function name : add
# arg : n
# return : n+10
```

```python
add= lambda n : n+10
add
```

```python
add(100)
```

```python
# def cube(a):
#     return(a*a*a)

cube= lambda a : a*a*a
cube(10)
```

**case-2: more than one argument**

```python
def summ(n1,n2):
    return(n1+n2)

def avg(n1,n2,n3):
    return((n1+n2+n3)/3)

def avg(a,b,c):
    return(round((a+b+c)/3))
```

```python
summ= lambda n1,n2 : n1+n2
summ(100,200)
```

```python
avg= lambda n1,n2,n3 : (n1+n2+n3)/3
avg(1,11,22)
```

```python
avg= lambda n1,n2,n3 : round((n1+n2+n3)/3,2)
avg(1,11,22)
```

```python
avg= lambda n1,n2,n3=100 : (n1+n2+n3)/3
avg(1,11)
```

**Case-3: if-else**

```python
<fun name>= lambda <arg> : <expression>
<fun name>= lambda <arg1>,<arg2> : <expression>
<fun name>= lambda <arg> : <if_out> <if con> else <else_out>

```

```python
even_odd= lambda i : 'even' if i%2==0 else 'odd'
even_odd(21)
```

```python
even_odd= lambda i : f'even_{i}' if i%2==0 else f'odd_{i}'
even_odd(21)
```

```python
# wap take two numbers find greater number using lambda

greater= lambda n1,n2 : n1 if n1>n2 else n2
greater(100,200)
```

```python
def Grade(n):
    if n>=75:
        return('A')
    elif n>=60:
        return('B')
    else:
        return('C')
Grade(67)
```

```python
Grade= lambda i : 'A' if i>=75 else \
                  'B' if i>=60 else \
                  'C'
Grade(21)
```

```python
Grade= lambda i : ('A' if i>=75 else 
                  'B' if i>=60 else 
                  'C')
Grade(21)
```
