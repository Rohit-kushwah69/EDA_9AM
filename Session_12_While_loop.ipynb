## While loop

- while loop mainly used for unlimited executions

- for examples we don't know when to stop the loop

- we don't know the condition to stop

- we have requirement the loop will run unlimted till we get True value then go for while

- if we know how many times the loop can run go with **for loop**

- while loop works for limited executions like for loop,also umlimited executions                                                      

- intial point

- condition to stop

- increment or decrement

```python
import random
random_num=random.randint(1,10)
user_num=eval(input('enter num:'))
if user_num==random_num:
    print('won')
else:
    print('False')
```

```python
for i in range(3):
    random_num=random.randint(1,10)
    user_num=eval(input('enter num:'))
    if user_num==random_num:
        print('won')
    else:
        print('False')
```

```python
i=1
while i<=3:
    random_num=random.randint(1,10)
    user_num=eval(input('enter num:'))
    if user_num==random_num:
        print('won')
    else:
        print('False')
    i=i+1
```

```python
i=1
while True:
    random_num=random.randint(1,10)
    print(random_num)
    user_num=eval(input('enter num:'))
    if user_num==random_num:
        print('won')
        break
    else:
        print('False')
    i=i+1
```

```python
while True:                          # infinite loop start
    password = input("Password: ")
    if password == "12345":      # stopping condition
        print("Welcome!")
        break                        # exit loop
    else:
        print("Try again")
```
