```python
# even and odd
# wap ask the user get 5 random numbers 
# print even or odd ==== save in a list
import random
l=[]
for i in range(5):
    num=random.randint(1,100)
    if num%2==0:
        #print(f'{num} is an even')
        l.append(num)
    else:
        #print(f'{num} is an odd')
        l.append(num)
l
```

**pop vs remove**

```python
l=['Apple','Banana','Cherry','Milk','Tea']
l.pop()
# Remove and return item at index (default last).
```

```python
l
```

```python
l=['Apple','Banana','Cherry','Milk','Tea','Banana']
# I want to delete banana
ID=l.index('Banana')
l.pop(ID)
l
```

```python
l=['Apple','Banana','Cherry','Milk','Tea','Banana']
l.remove('Banana')
l
```

- pop takes index

- remove takes direct element name

    - taking elemnet name advantage

    - only first element only drop

- when we want to drop second element use pop method
