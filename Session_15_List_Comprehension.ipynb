**List Comprehesnion**

- List comprehension means writing a code in a single line with out using append

- Many codes having multiple lines with conditions and loops slower the execution

- List Comprehension will help you speed up the process

$case-1$

- focus on what is the output variable

    - which means which variable we are saving


```python
l=[]
for i in range(10):
    l.append(i)
l
```

```python
# syntax
# [<output variable>   <for loop>]
l=[i for i in range(10)]
l
```

```python
# Q1) input=['hyd','bengaluru','mumbai','chennai']
#     output=['Hyd','Bengaluru','Mumbai','Chennai']
# Q2) input=['hyd','bengaluru','mumbai','chennai']
#     output=['HYD','BENGALURU','MUMBAI','CHENNAI']
Input=['hyd','bengaluru','mumbai','chennai']
output=[]
for i in Input:
    output.append(i.capitalize())

output=[i.capitalize() for i in Input]
```

```python
Input=['hyd','bengaluru','mumbai','chennai']
output=[i.capitalize() for i in Input]
output
```

```python
Input=[20,21,30,31,34]
even=[]
for i in Input:
    if i%2==0:
        even.append(i)
# output : i
# for i in Input
# if i%2==0
```

```python
Input=[20,21,30,31,34]
even=[i for i in Input if i%2==0]
odd=[i for i in Input if i%2!=0]
odd
```

$Case-3$: **for-if-else**

```python
for i in Input:
    if '#' in i:
        print(i)
    else:
        print(i)
```

```python
[<output> <for loop>]
[<output> <for loop> <if condition>]
[<if_out> <if_con> else <else out> <for loop>] 
```

```python
Input=[20,21,30,31,34]
out=[]
for i in Input:
    if i%2==0:
        out.append(f'even_{i}')
    else:
        out.append(f'odd_{i}')
out
```

```python
# if_out:f'even_{i}'
# if_con : if i%2==0
# else_out: f'odd_{i}'
# for i in Input

[f'even_{i}' if i%2==0 else f'odd_{i}' for i in Input]
```

$Case-4$:**if-elif-else**

```python
[<output> <for loop>]
[<out> <for loop> <if cond>]
[<if out> <if cond> else <else out> <for loop>]
[<if out> <if cond> else <elif out> <elif cond> else <else out> <for loop>]
# Rule-1: there is no elif
          # else if 
# Rule-2: if after  else
# Rule-3: else after output

Input=[50,70,90] # C  B  A
out=[]
for i in Input:
    if i>=90:
        out.append('A')
    elif i>=60:
        out.append('B')
    else:
        out.append('C')

```

```python
# if_out if_cond \
# else elif_out elif_cond \
# else elif_out elif_cond \
# else elif_out elif_cond \
# else else_out
Input=[50,70,90]

['A' if i>=90 \
else 'B' if i>=60 \
else 'C' \
for i in Input]
```

```python
# [<out> <for loop>]
# <out> <for> <if>
# [<if out> <if cond> else <els out> <for loop>]
# <if out> <if> else <elif out> <elif con> else <else out> <>
```
